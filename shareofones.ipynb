```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# скорость в месяцах (как непрерывная величина)
dealsdf['скорость_продажи_мес'] = (
    (dealsdf['дата_договора_(месяц.год)'] - dealsdf['дата_начала_продаж'])
    / np.timedelta64(1, 'D')
) / 30.4375


speed_df = (
    dealsdf
    .groupby('id_корпуса', as_index=False)
    .agg(
        speed_months=('скорость_продажи_мес', 'median')
    )
)

total_deals = (
    dealsdf
    .groupby('id_корпуса')
    .size()
    .reset_index(name='total_deals')
)

one_st_deals = (
    dealsdf[dealsdf['количество_комнат'].isin(['1', 'ст'])]
    .groupby('id_корпуса')
    .size()
    .reset_index(name='one_st_deals')
)

share_df = (
    total_deals
    .merge(one_st_deals, on='id_корпуса', how='left')
    .fillna({'one_st_deals': 0})
)

share_df['share_1_st'] = share_df['one_st_deals'] / share_df['total_deals']

print(share_df)

df_plot = speed_df.merge(share_df, on='id_корпуса', how='inner')

fig, ax = plt.subplots(figsize=(16, 9))

ax.scatter(
    df_plot['share_1_st'],
    df_plot['speed_months'],
    alpha=0.6
)

ax.set_xlabel('Доля 1-комн + студий')
ax.set_ylabel('Скорость продажи (мес от старта продаж)')
ax.set_title('Зависимость скорости продаж от доли 1-комн и студий')
ax.grid(True, alpha=0.3)

plt.show()

```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(16, 9))

import numpy as np

x = df_plot['share_1_st'].to_numpy()
y = df_plot['speed_months'].to_numpy()

# коэффициенты y = a*x + b
a, b = np.polyfit(x, y, 1)

# предсказания
y_pred = a * x + b

# R^2
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y.mean()) ** 2)
r2 = 1 - ss_res / ss_tot


# scatter
ax.scatter(
    x,
    y,
    alpha=0.6,
    label='Корпуса'
)

# линия регрессии
x_line = np.linspace(x.min(), x.max(), 200)
y_line = a * x_line + b

ax.plot(
    x_line,
    y_line,
    linewidth=2,
    label=f'Линейная регрессия: y = {a:.2f}x + {b:.2f}\nR² = {r2:.3f}'
)

ax.set_xlabel('Доля 1-комн + студий')
ax.set_ylabel('Медианная скорость продаж (мес от старта)')
ax.set_title('Линейная зависимость скорости продаж от доли 1-комн и студий')
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```

```python
from scipy.stats import ttest_ind

df = df_plot.copy()

# порог — медиана
threshold = df['share_1_st'].median()

df['group'] = np.where(
    df['share_1_st'] >= threshold,
    'high_share',
    'low_share'
)

high = df.loc[df['group'] == 'high_share', 'speed_months']
low  = df.loc[df['group'] == 'low_share',  'speed_months']

print(f'High share n = {len(high)}')
print(f'Low  share n = {len(low)}')

summary = df.groupby('group')['speed_months'].agg(
    mean='mean',
    median='median',
    std='std'
)

print(summary)

t_stat, p_value = ttest_ind(
    high,
    low,
    equal_var=False,   # ← ключевое отличие Уэлча
    nan_policy='omit'
)

print(f't-statistic = {t_stat:.3f}')
print(f'p-value     = {p_value:.4f}')

def cohens_d(x, y):
    nx, ny = len(x), len(y)
    vx, vy = x.var(ddof=1), y.var(ddof=1)
    pooled_std = np.sqrt((vx + vy) / 2)
    return (x.mean() - y.mean()) / pooled_std

d = cohens_d(high, low)
print(f"Cohen's d = {d:.3f}")

q30 = df['share_1_st'].quantile(0.05)
q70 = df['share_1_st'].quantile(0.95)

low_q  = df.loc[df['share_1_st'] <= q30, 'speed_months']
high_q = df.loc[df['share_1_st'] >= q70, 'speed_months']

print(f'High share n = {len(high_q)}')
print(f'Low  share n = {len(low_q)}')

t_stat, p_value = ttest_ind(high_q, low_q, equal_var=False)
print(f't-statistic = {t_stat:.3f}')
print(f'p-value     = {p_value:.4f}')

d = cohens_d(high_q, low_q)
print(f"Cohen's d = {d:.3f}")
```

```python
from sklearn.linear_model import HuberRegressor
import numpy as np

X = df_plot[['share_1_st']].values
y = df_plot['speed_months'].values

model = HuberRegressor()
model.fit(X, y)

beta_hat = model.coef_[0]
print(f'β̂ = {beta_hat:.4f}')

rng = np.random.default_rng(42)
B = 2000
betas = []

n = len(df_plot)

for _ in range(B):
    idx = rng.choice(n, size=n, replace=True)
    Xb = X[idx]
    yb = y[idx]
    m = HuberRegressor().fit(Xb, yb)
    betas.append(m.coef_[0])

betas = np.array(betas)

# односторонний p-value
p_value = np.mean(betas >= 0)

print(f'Bootstrap p-value (H₁: β < 0) = {p_value:.4f}')

import matplotlib.pyplot as plt

# линия регрессии
x_line = np.linspace(X.min(), X.max(), 300).reshape(-1, 1)
y_line = model.predict(x_line)

# график
fig, ax = plt.subplots(figsize=(16, 9))

ax.scatter(
    X.flatten(),
    y,
    alpha=0.6,
    label='Корпуса'
)

ax.plot(
    x_line,
    y_line,
    linewidth=3,
    label='Robust линейная регрессия (Huber)'
)

ax.set_xlabel('Доля 1-комн + студий')
ax.set_ylabel('Скорость продажи (мес от старта продаж)')
ax.set_title(
    'Зависимость скорости продаж от доли 1-комн и студий\n'
    'Robust линейная регрессия'
)

ax.grid(True, alpha=0.3)
ax.legend()

# подпись с результатами
ax.text(
    0.02, 0.98,
    f'β̂ = {beta_hat:.2f}\nBootstrap p < 0.001',
    transform=ax.transAxes,
    va='top',
    bbox=dict(boxstyle='round', facecolor='white', alpha=0.85)
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# (опционально) seaborn для красивой heatmap
try:
    import seaborn as sns
    _HAS_SNS = True
except ImportError:
    _HAS_SNS = False


# =========================
# 1) Подготовка данных
# =========================

df = dealsdf.copy()

# даты
df['дата_договора_(месяц.год)'] = pd.to_datetime(df['дата_договора_(месяц.год)'], errors='coerce')
df['дата_начала_продаж'] = pd.to_datetime(df['дата_начала_продаж'], errors='coerce')

# скорость продажи (мес от старта продаж)
df['speed_months'] = (
    (df['дата_договора_(месяц.год)'] - df['дата_начала_продаж']) / np.timedelta64(1, 'D')
) / 30.4375

# комнаты как категория
df['rooms_cat'] = (
    df['количество_комнат']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({'ст': 'студия', 'studio': 'студия'})
    .replace({'1.0': '1', '2.0': '2', '3.0': '3', '4.0': '4'})
)

room_order = ['студия', '1', '2', '3', '4']  # оставь только те, что реально есть
df = df[df['rooms_cat'].isin(room_order)].copy()

df['rooms_cat'] = pd.Categorical(df['rooms_cat'], categories=room_order, ordered=True)

# фильтры от мусора (можешь подстроить)
df = df.dropna(subset=['speed_months', 'rooms_cat'])
df = df[(df['speed_months'] >= 0) & (df['speed_months'] <= 60)]  # отсечь совсем экстремальные

# =========================
# 2) Риск (порог по плану)
# =========================
T = 18  # плановый срок, мес
df['slow'] = (df['speed_months'] > T).astype(int)

# =========================
# 3) Бинирование скорости (ось Y)
# =========================
speed_bins = [0, 3, 6, 9, 12, 18, np.inf]
speed_labels = ['<3', '3–6', '6–9', '9–12', '12–18', '>18']

df['speed_bin'] = pd.cut(df['speed_months'], bins=speed_bins, labels=speed_labels, include_lowest=True)

# =========================
# 4) Heatmap: вероятность "slow" по rooms x speed_bin
# =========================
risk_by_rooms = (
    df.groupby('rooms_cat', observed=True)['slow']
      .agg(
          risk_prob='mean',
          n='size',
          median_speed='median'
      )
)

print(risk_by_rooms)

fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(
    risk_by_rooms.index.astype(str),
    risk_by_rooms['risk_prob']
)

ax.set_ylabel(f'P(speed_months > {T})')
ax.set_xlabel('Тип квартиры')
ax.set_title('Риск медленных продаж по типу квартиры')

ax.grid(True, axis='y', alpha=0.3)
plt.show()
```

```python
df = dealsdf.copy()

# даты
df['дата_договора_(месяц.год)'] = pd.to_datetime(df['дата_договора_(месяц.год)'], errors='coerce')
df['дата_начала_продаж'] = pd.to_datetime(df['дата_начала_продаж'], errors='coerce')

# скорость продажи (мес)
df['speed_months'] = (
    (df['дата_договора_(месяц.год)'] - df['дата_начала_продаж'])
    / np.timedelta64(1, 'D')
) / 30.4375

df = df.dropna(subset=['speed_months', 'этаж_лота'])
df['этаж_лота'] = pd.to_numeric(df['этаж_лота'], errors='coerce')
df = df.dropna(subset=['этаж_лота'])

# риск = превышение плана
T = 12
df['slow'] = (df['speed_months'] > T).astype(int)

df['floor_bin'] = pd.cut(
    df['этаж_лота'],
    bins=[0, 2, 5, 10, 15, np.inf],
    labels=['1–2', '3–5', '6–10', '11–15', '16+']
)

risk_by_floor = (
    df.groupby('floor_bin', observed=True)['slow']
      .agg(
          risk_prob='mean',
          n='size',
          median_speed='median'
      )
)

print(risk_by_floor)

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    risk_by_floor.index.astype(str),
    risk_by_floor['risk_prob'],
    marker='o',
    color="#FFD700",
)

ax.set_ylabel(f'P(speed_months > {T})')
ax.set_xlabel('Этаж')
ax.set_title('Риск медленных продаж по этажам')
ax.grid(True, alpha=0.3)

plt.show()
```

```python
from scipy.stats import spearmanr

# берём исходные этажи (не бины)
tmp = df[['этаж_лота', 'slow']].dropna()

rho, p_two_sided = spearmanr(tmp['этаж_лота'], tmp['slow'])

# односторонний p-value (H₁: rho > 0)
p_one_sided = p_two_sided / 2 if rho > 0 else 1.0

print(f"Spearman rho = {rho:.3f}")
print(f"one-sided p-value = {p_one_sided:.4f}")
```

```python
low = df.loc[df['floor_bin'].isin(['1–2','3–5']), 'slow']
high = df.loc[df['floor_bin'].isin(['11–15','16+']), 'slow']

from scipy.stats import mannwhitneyu

u = mannwhitneyu(high, low, alternative='greater')
print(f"Mann–Whitney p-value = {u.pvalue:.4f}")
```
