```python
metronew = pd.read_csv("newmetro.csv")
metronew
```

```python
import re

import re

def dms_to_decimal(dms):
    nums = list(map(float, re.findall(r"\d+", dms)))
    deg, minute, sec = nums[:3]
    return deg + minute / 60 + sec / 3600

metronew["название"] = (
    metronew["Название станции\nПрежние названия"]
    .str.split("\n")
    .str[0]
    .str.replace("«|»", "", regex=True)
)

def extract_lat_lon(s):
    lat_dms = re.search(r"(\d+°\d+′\d+″)\s*с\.?\s*ш\.?", s)
    lon_dms = re.search(r"(\d+°\d+′\d+″)\s*в\.?\s*д\.?", s)

    lat = dms_to_decimal(lat_dms.group(1)) if lat_dms else None
    lon = dms_to_decimal(lon_dms.group(1)) if lon_dms else None

    return pd.Series([lat, lon])

metronew[["lat", "lon"]] = metronew["Координаты"].apply(extract_lat_lon)
metronew
```

```python
finalmetro = metronew[["название", "lat", "lon"]]
finalmetro
```

```python
import numpy as np
import pandas as pd

def haversine_km(lat1, lon1, lat2, lon2):
    # lat1/lon1: (n,1), lat2/lon2: (1,m)
    R = 6371.0088
    lat1 = np.deg2rad(lat1); lon1 = np.deg2rad(lon1)
    lat2 = np.deg2rad(lat2); lon2 = np.deg2rad(lon2)
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

# на всякий случай приводим к числам
output["lat"] = pd.to_numeric(output["lat"], errors="coerce")
output["lon"] = pd.to_numeric(output["lon"], errors="coerce")
finalmetro["lat"] = pd.to_numeric(finalmetro["lat"], errors="coerce")
finalmetro["lon"] = pd.to_numeric(finalmetro["lon"], errors="coerce")

# матрица расстояний n x m
lat_o = output["lat"].to_numpy()[:, None]
lon_o = output["lon"].to_numpy()[:, None]
lat_m = finalmetro["lat"].to_numpy()[None, :]
lon_m = finalmetro["lon"].to_numpy()[None, :]

dist_km = haversine_km(lat_o, lon_o, lat_m, lon_m)

# индекс ближайшего метро для каждой строки output
idx = np.nanargmin(dist_km, axis=1)

output["ближайшее_метро"] = finalmetro["название"].to_numpy()[idx]
output["расстояние_км"] = dist_km[np.arange(dist_km.shape[0]), idx]
output["расстояние_м"] = (output["расстояние_км"] * 1000).round(0).astype("Int64")

output
```
